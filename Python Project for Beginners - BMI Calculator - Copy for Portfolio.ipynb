```python
# Request the user's name
name = input("Please enter your name: ") 

# Request the user's weight in pounds
weight = int(input("Please enter your weight in pounds: "))

# Request the user's height in inches
height = int(input("Please enter your height in inches: "))

# Calculate the user's BMI based off of the above numbers; 
# round that value to the nearst one decimal place
# Formula taken from the following source:
# https://www.cdc.gov/growth-chart-training/hcp/using-bmi/body-mass-index.html
BMI = round((weight / (height ** 2) * 703), 1)

# Output the user's BMI. 
print("Your BMI is " + str(BMI) + ".")

# Provide some context to the user
# Note: Weight classifications had been taken from the following source:
# https://www.qlifefamilyclinic.com/blog-detail/31/obesity
if (0 < BMI): 
    if (BMI < 18.5):
        print(name + ", it appears that you are underweight.")
    elif (BMI <= 24.9):
        print(name + ", it appears that you are normal weight.")
    elif (BMI <= 29.9):
        print(name + ", it appears that you are overweight.")
    elif (BMI <= 34.9):
        print(name + ", it appears that you are class I obese.")
    elif (BMI <= 39.9):
        print(name + ", it appears that you are class II obese.")
    else:
        print(name + ", it appears that you are class III obese.")
else: 
    print("It appears that one of your inputs is invalid.")
```
